## Day 15 – Analysis & Communication

### Observations (3-5 sentences)
*Focus: Describe measurable results only — no explanation.*

> The quantitative evaluation shows that filtering RMSE increases monotonically with kernel size, from the smallest value at M=3 to the largest at M=20. Peak Preservation Ratio decreases with increasing kernel size, confirming progressive attenuation of extreme values. Gaussian filtering at M=5 produces slightly lower RMSE and higher peak preservation than the corresponding moving average, indicating better feature retention. Wiener deconvolution with λ=0.01 achieves the lowest reconstruction RMSE among tested regularization values, while λ=0.001 amplifies noise and λ=1.0 produces over-smoothed reconstructions. The reconstructed signal at optimal λ closely approximates the original in the interior but shows boundary artifacts.

### Interpretation (3-5 sentences)
*Focus: Explain what the results mean for the application.*

> In the context of SMI financial risk monitoring, these results imply that moderate smoothing (M=5, corresponding to 1 trading week) provides a practical balance between noise reduction and feature preservation. The observed decrease in peak preservation with larger kernels directly affects the detection of drawdowns and trend reversals, because smoothing delays and attenuates these critical risk indicators. Wiener deconvolution can meaningfully improve signal interpretation when the regularization parameter is appropriately tuned (λ ≈ 0.01), but becomes counterproductive at extreme values. This suggests that convolution-based smoothing with Gaussian kernels is particularly effective for the objective defined in the problem statement: preserving temporal sharpness while reducing short-term noise.

### Discussion and Critical Reflection (4-6 sentences)
*Focus: Relate the quantitative findings to the requirements of the defined use case and assess practical adequacy.*

> For the defined use case, Gaussian smoothing with M=5 performed well because it achieved meaningful noise reduction while maintaining high peak preservation, which aligns with the requirement of timely risk detection. In contrast, aggressive smoothing (M=20) showed reduced performance, leading to significant peak attenuation and delayed detection of trend reversals, limiting its suitability for intraday or weekly risk monitoring. Wiener deconvolution at λ=0.01 achieved reconstruction quality that meets the practical expectations defined in the methodological design, recovering most of the original signal's dynamic range. The implementation assumes that the smoothing kernel is perfectly known and that additive noise is white, which may be constrained under real-world conditions such as non-stationary market dynamics or imperfect knowledge of the applied filter. To improve robustness, future work should address adaptive kernel selection that adjusts smoothing strength based on local signal variability, or refine the noise model used in Wiener deconvolution.
