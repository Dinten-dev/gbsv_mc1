## Day 5 – Analysis & Communication

### Observations (3-5 sentences)
*Focus: Describe measurable results only — no explanation.*

> The quantitative evaluation shows that RMSE increases monotonically from 0.00 CHF at D=1 (baseline) to progressively larger values as the downsampling factor increases. The relative volatility error similarly rises with increasing D. Performance differences are most pronounced between D=5 and D=60, where RMSE and volatility error increase steeply. The frequency spectra confirm spectral energy redistribution and aliasing artifacts become visually dominant for D ≥ 5. At D=2 (Nyquist boundary), reconstruction error remains modest, indicating marginal but acceptable fidelity.

### Interpretation (3-5 sentences)
*Focus: Explain what the results mean for the application.*

> In the context of SMI intraday financial risk monitoring, these results imply that sampling intervals beyond 5 minutes introduce substantial distortion in both amplitude fidelity and volatility estimation. The observed increase in relative volatility error affects the accuracy of risk metrics such as Value-at-Risk (VaR) and realized volatility, because undersampled signals systematically miss intraday extreme movements. This suggests that maintaining a sampling interval at or below 2 minutes (D ≤ 2) is particularly critical for achieving the objective defined in the problem statement: preserving the spectral content of intraday price dynamics for reliable volatility estimation.

### Discussion and Critical Reflection (4-6 sentences)
*Focus: Relate the quantitative findings to the requirements of the defined use case and assess practical adequacy.*

> For the defined use case, the configurations D=1 and D=2 performed well because they achieved low RMSE and negligible volatility estimation error, which aligns with the requirement of accurate intraday risk assessment. In contrast, configurations D=15 and D=60 showed reduced performance, leading to significant amplitude distortion and volatility underestimation, limiting their suitability for real-time market surveillance. The achieved performance level at D ≤ 2 can be considered sufficient for the use case because the dominant market dynamics occur on time scales of minutes or longer, and the Nyquist criterion is satisfied. The implementation and evaluation approach assumes quasi-stationarity of the price signal, which may be constrained under real-world conditions such as flash crashes, trading halts, or central bank announcements that introduce non-stationary broadband transients. To improve robustness or applicability, future work should address adaptive sampling strategies that increase temporal resolution during periods of elevated market activity, or refine the anti-aliasing filter design before downsampling.
