## Day 7 – Methodological Design

### Theoretical Foundation and Method Choice
*Focus: principled justification aligned with the use case*

> This investigation applies the principle of **statistical correlation analysis** to the SMI daily closing prices within the context of market cycle detection and pattern recognition in Swiss financial markets.

**Auto-correlation** is defined as:
$$R_{xx}[\tau] = \frac{1}{N} \sum_{n=0}^{N-1-\tau} (x[n] - \bar{x})(x[n+\tau] - \bar{x})$$

When normalized by $R_{xx}[0]$, the auto-correlation function (ACF) produces values in $[-1, 1]$. Peaks in the ACF at lag $\tau$ indicate that the signal repeats with period $\tau$.

**Cross-correlation** (normalized) is defined as:
$$\rho_{xy}[\tau] = \frac{\sum_{n} (x[n] - \bar{x})(y[n+\tau] - \bar{y})}{\sqrt{\sum_n (x[n] - \bar{x})^2 \cdot \sum_n (y[n+\tau] - \bar{y})^2}}$$

This measures the similarity between a short template $x$ and a sliding window of the signal $y$ at each offset $\tau$.

> This principle assumes stationarity or quasi-stationarity and is valid when periodic structures or repeating patterns actually exist in the data. Auto-correlation is selected for global periodicity detection because it identifies dominant repeating cycles without requiring prior knowledge of their period. Cross-correlation is selected for localized segment matching because it directly quantifies shape similarity between a template and signal segments. If the stationarity assumption is violated — for example, during structural market regime changes — spurious correlation peaks may appear, compromising pattern detection reliability in this application.

### Parameter Definition and Mathematical Specification
*Focus: explicit parameter selection, derivation, and unit consistency*

> The signal is characterized by a sampling interval of 1 trading day, N ≈ 250 data points, expressed in CHF.

**Auto-correlation parameters:**
- Maximum lag: $\tau_{\max} = 60$ trading days (~3 months), justified because the expected dominant periodicities (weekly ~5 days, monthly ~22 days) fall well within this range
- Normalization: by $R_{xx}[0]$ to produce normalized ACF in $[-1, 1]$
- 95% confidence band for white noise: $\pm 1.96 / \sqrt{N}$, used to assess statistical significance of ACF peaks

**Cross-correlation parameters:**
- Template length: 10 trading days (~2 weeks), chosen as a balance between capturing meaningful price patterns and maintaining localization precision
- Template origin: extracted from a visually identified local minimum/recovery segment
- Normalization: normalized cross-correlation (NCC, range $[-1, 1]$) to make comparison amplitude-independent

### Experimental Design for Next Days
*Focus: structured parameter variation and theoretical prediction*

> The baseline configuration is defined as: normalized auto-correlation with $\tau_{\max} = 60$, and normalized cross-correlation with template length $L = 10$.

The following parameters will be varied systematically:

| Parameter | Range | Justification |
|-----------|-------|---------------|
| Max lag (auto-corr.) | 30, 60, 120 | Tests whether shorter/longer lag windows capture or miss dominant cycles |
| Template length (cross-corr.) | 5, 10, 20 | Tests localization vs. selectivity trade-off |
| Noise level (robustness) | SNR = ∞, 20 dB, 10 dB | Tests cross-correlation under realistic perturbation |
| Amplitude scaling | 1.0, 0.8, 1.2 | Tests NCC robustness to amplitude variation |

> It is theoretically expected that increasing max lag reveals longer-period cycles but reduces statistical reliability at high lags. Shorter templates increase localization but reduce selectivity. Additive noise degrades cross-correlation peak sharpness. NCC should be inherently robust to amplitude scaling.

### Methodological Limitations and Risk Factors
*Focus: assumptions, stability, and potential misinterpretation*

> This approach assumes quasi-stationarity of the SMI price signal, which may be violated during market regime changes (e.g., crisis onset, central bank interventions).  
> The method is expected to be reliable when dominant periodic structures are present and the signal length substantially exceeds the periods of interest, but may become unstable or misleading when trend components dominate and mask oscillatory behavior.  
> In this financial analysis use case, the primary risk factors are non-stationarity and trend contamination, potentially leading to slowly decaying ACF that reflects persistence rather than true periodicity.
